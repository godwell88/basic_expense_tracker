```python

import tkinter as tk
from tkinter import messagebox

# Create the main window
root = tk.Tk()
root.title("Monthly Expense Tracker")
root.geometry("450x650")

# Store budget information
monthly_income = 0
savings_goal = 0
total_expenses = 0


# Set the monthly budget
def set_budget():
    global monthly_income, savings_goal, total_expenses

    try:
        income = float(income_entry.get())
        savings = float(savings_entry.get())

        if income <= 0:
            messagebox.showerror("Error", "Income must be greater than zero")
            return

        if savings < 0:
            messagebox.showerror("Error", "Savings cannot be negative")
            return

        if savings > income:
            messagebox.showerror(
                "Error",
                "Your savings goal cannot exceed your monthly income"
            )
            return

        monthly_income = income
        savings_goal = savings

        # Reset expenses when setting a new budget
        total_expenses = 0
        expense_list.delete(0, tk.END)

        available = monthly_income - savings_goal

        income_label.config(
            text=f"Monthly Income: ${monthly_income:.2f}"
        )

        savings_label.config(
            text=f"Savings Goal: ${savings_goal:.2f}"
        )

        update_balance()

        messagebox.showinfo(
            "Success",
            f"Budget set successfully!\n"
            f"You can spend ${available:.2f} this month."
        )

    except ValueError:
        messagebox.showerror(
            "Error",
            "Please enter valid numbers for income and savings"
        )


# Add a new expense
def add_expense():
    global total_expenses

    if monthly_income <= 0:
        messagebox.showerror(
            "Error",
            "Please set your monthly budget first"
        )
        return

    name = name_entry.get().strip()
    category = category_entry.get().strip()
    amount_text = amount_entry.get().strip()

    if not name or not category or not amount_text:
        messagebox.showerror("Error", "Please fill in all expense fields")
        return

    try:
        amount = float(amount_text)

        if amount <= 0:
            messagebox.showerror(
                "Error",
                "Expense amount must be greater than zero"
            )
            return

        available = monthly_income - savings_goal - total_expenses

        if amount > available:
            messagebox.showerror(
                "Budget Warning",
                f"You only have ${available:.2f} left to spend."
            )
            return

        total_expenses += amount

        expense_list.insert(
            tk.END,
            f"{name} - {category} - ${amount:.2f}"
        )

        update_balance()

        # Clear the expense fields
        name_entry.delete(0, tk.END)
        amount_entry.delete(0, tk.END)
        category_entry.delete(0, tk.END)

    except ValueError:
        messagebox.showerror(
            "Error",
            "Please enter a valid expense amount"
        )


# Update the remaining budget
def update_balance():
    available = monthly_income - savings_goal
    remaining = available - total_expenses

    total_label.config(
        text=f"Total Expenses: ${total_expenses:.2f}"
    )

    remaining_label.config(
        text=f"Money Left to Spend: ${remaining:.2f}"
    )


# Main heading
tk.Label(
    root,
    text="MONTHLY EXPENSE TRACKER",
    font=("Arial", 16, "bold")
).pack(pady=10)


# Monthly income section
tk.Label(root, text="Enter Monthly Income").pack()

income_entry = tk.Entry(root, width=30)
income_entry.pack(pady=5)


# Savings goal section
tk.Label(root, text="How Much Do You Want to Save?").pack()

savings_entry = tk.Entry(root, width=30)
savings_entry.pack(pady=5)

tk.Button(
    root,
    text="Set Monthly Budget",
    command=set_budget
).pack(pady=10)


# Budget summary
income_label = tk.Label(root, text="Monthly Income: $0.00")
income_label.pack()

savings_label = tk.Label(root, text="Savings Goal: $0.00")
savings_label.pack()

tk.Label(
    root,
    text="ADD YOUR EXPENSES",
    font=("Arial", 12, "bold")
).pack(pady=10)


# Expense name
tk.Label(root, text="Expense Name").pack()

name_entry = tk.Entry(root, width=30)
name_entry.pack(pady=3)


# Expense amount
tk.Label(root, text="Expense Amount").pack()

amount_entry = tk.Entry(root, width=30)
amount_entry.pack(pady=3)


# Expense category
tk.Label(root, text="Expense Category").pack()

category_entry = tk.Entry(root, width=30)
category_entry.pack(pady=3)


# Add expense button
tk.Button(
    root,
    text="Add Expense",
    command=add_expense
).pack(pady=10)


# Expense list
tk.Label(
    root,
    text="YOUR EXPENSES",
    font=("Arial", 12, "bold")
).pack(pady=5)

expense_list = tk.Listbox(root, width=45, height=8)
expense_list.pack(pady=5)


# Budget totals
total_label = tk.Label(
    root,
    text="Total Expenses: $0.00",
    font=("Arial", 11, "bold")
)
total_label.pack(pady=5)

remaining_label = tk.Label(
    root,
    text="Money Left to Spend: $0.00",
    font=("Arial", 12, "bold")
)
remaining_label.pack(pady=5)


# Run the application
root.mainloop()
```

```python

# Monthly Expense Tracker

# Get monthly income and savings goal
monthly_income = float(input("Enter your monthly income: $"))
savings_goal = float(input("How much do you want to save this month? $"))

# Validate budget
if monthly_income <= 0 or savings_goal < 0 or savings_goal > monthly_income:
    print("Invalid income or savings goal.")
else:
    available_budget = monthly_income - savings_goal
    total_expenses = 0
    expenses = []

    # Available categories
    categories = ["Food", "Personal Expenses", "Accommodation"]

    print("\n===================================")
    print("       MONTHLY EXPENSE TRACKER")
    print("===================================")
    print(f"Monthly Income:       ${monthly_income:.2f}")
    print(f"Savings Goal:         ${savings_goal:.2f}")
    print(f"Available to Spend:   ${available_budget:.2f}")

    while True:
        print("\n1. Add Expense")
        print("2. View Expenses")
        print("3. View Budget Summary")
        print("4. Exit")

        choice = input("\nChoose an option: ")

        # Add expense
        if choice == "1":
            name = input("Enter expense name: ")

            print("\nChoose a category:")
            for i, category in enumerate(categories, start=1):
                print(f"{i}. {category}")

            category_choice = input("Enter category number: ")

            if category_choice not in ["1", "2", "3"]:
                print("Invalid category. Please try again.")
                continue

            category = categories[int(category_choice) - 1]

            try:
                amount = float(input("Enter expense amount: $"))

                if amount <= 0:
                    print("Amount must be greater than zero.")
                    continue

                remaining = available_budget - total_expenses

                if amount > remaining:
                    print(f"Not enough money available!")
                    print(f"You can spend only ${remaining:.2f}.")
                    continue

                # Store expense
                expense = {
                    "name": name,
                    "category": category,
                    "amount": amount
                }

                expenses.append(expense)
                total_expenses += amount

                print("Expense added successfully!")

            except ValueError:
                print("Please enter a valid amount.")

        # View expenses in a table
        elif choice == "2":
            if not expenses:
                print("\nNo expenses recorded yet.")
            else:
                print("\n" + "=" * 65)
                print(f"{'Expense Name':<25}{'Category':<25}{'Amount':>12}")
                print("-" * 65)

                for expense in expenses:
                    print(
                        f"{expense['name']:<25}"
                        f"{expense['category']:<25}"
                        f"${expense['amount']:>10.2f}"
                    )

                print("-" * 65)
                print(f"{'Total Expenses':<50}${total_expenses:>10.2f}")
                print("=" * 65)

        # View budget summary
        elif choice == "3":
            remaining = available_budget - total_expenses

            print("\n========== BUDGET SUMMARY ==========")
            print(f"Monthly Income:       ${monthly_income:.2f}")
            print(f"Savings Goal:         ${savings_goal:.2f}")
            print(f"Available to Spend:   ${available_budget:.2f}")
            print(f"Total Expenses:       ${total_expenses:.2f}")
            print(f"Money Left to Spend:  ${remaining:.2f}")
            print("====================================")

        # Exit
        elif choice == "4":
            print("\nThank you for using the Expense Tracker!")
            break

        else:
            print("Invalid option. Please choose 1, 2, 3, or 4.")
```

```python
class ExpenseTracker:
    def __init__(self):
        self.income = 0
        self.savings_goal = 0
        self.expenses = []

        self.categories = [
            "Food",
            "Personal Expenses",
            "Accommodation"
        ]

    # Set monthly income and savings goal
    def set_budget(self):
        while True:
            try:
                self.income = float(input("Enter your monthly income: $"))

                if self.income <= 0:
                    print("Income must be greater than zero.")
                    continue

                break

            except ValueError:
                print("Please enter a valid amount.")

        while True:
            try:
                self.savings_goal = float(
                    input("How much do you want to save this month? $")
                )

                if self.savings_goal < 0:
                    print("Savings cannot be negative.")
                    continue

                if self.savings_goal > self.income:
                    print("Savings cannot exceed your monthly income.")
                    continue

                break

            except ValueError:
                print("Please enter a valid amount.")

        print("\nBudget set successfully!")
        print(f"Monthly Income: ${self.income:.2f}")
        print(f"Savings Goal: ${self.savings_goal:.2f}")
        print(f"Available to Spend: ${self.available_budget():.2f}")

    # Calculate the available spending budget
    def available_budget(self):
        return self.income - self.savings_goal

    # Calculate total expenses
    def total_expenses(self):
        return sum(expense["amount"] for expense in self.expenses)

    # Calculate money remaining
    def remaining_balance(self):
        return self.available_budget() - self.total_expenses()

    # Add a new expense
    def add_expense(self):
        if self.income <= 0:
            print("\nPlease set your monthly budget first.")
            return

        print("\nChoose an expense category:")

        for number, category in enumerate(self.categories, start=1):
            print(f"{number}. {category}")

        try:
            choice = int(input("Enter category number: "))

            if choice < 1 or choice > len(self.categories):
                print("Invalid category selection.")
                return

            category = self.categories[choice - 1]

        except ValueError:
            print("Please enter a valid category number.")
            return

        name = input("Enter expense name: ").strip()

        if not name:
            print("Expense name cannot be empty.")
            return

        try:
            amount = float(input("Enter expense amount: $"))

            if amount <= 0:
                print("Expense amount must be greater than zero.")
                return

        except ValueError:
            print("Please enter a valid amount.")
            return

        if amount > self.remaining_balance():
            print(
                f"Not enough spending money! "
                f"You have ${self.remaining_balance():.2f} left."
            )
            return

        # Store the expense as a dictionary
        expense = {
            "name": name,
            "category": category,
            "amount": amount
        }

        self.expenses.append(expense)

        print("\nExpense added successfully!")

    # Display expenses in a table
    def view_expenses(self):
        if not self.expenses:
            print("\nNo expenses recorded yet.")
            return

        print("\n" + "=" * 65)
        print(
            f"{'No.':<5}"
            f"{'Expense Name':<25}"
            f"{'Category':<23}"
            f"{'Amount':>12}"
        )
        print("-" * 65)

        for number, expense in enumerate(self.expenses, start=1):
            print(
                f"{number:<5}"
                f"{expense['name']:<25}"
                f"{expense['category']:<23}"
                f"${expense['amount']:>10.2f}"
            )

        print("=" * 65)
        print(f"{'Total Expenses:':<50}${self.total_expenses():>10.2f}")

    # Display budget summary
    def view_budget(self):
        print("\n========== MONTHLY BUDGET ==========")
        print(f"Monthly Income:       ${self.income:.2f}")
        print(f"Savings Goal:         ${self.savings_goal:.2f}")
        print(f"Available to Spend:   ${self.available_budget():.2f}")
        print(f"Total Expenses:       ${self.total_expenses():.2f}")
        print(f"Money Left to Spend:  ${self.remaining_balance():.2f}")
        print("====================================")


# Main program
def main():
    tracker = ExpenseTracker()

    print("=================================")
    print("      MONTHLY EXPENSE TRACKER")
    print("=================================")

    tracker.set_budget()

    while True:
        print("\n1. Add Expense")
        print("2. View All Expenses")
        print("3. View Budget Summary")
        print("4. Exit")

        choice = input("Choose an option: ").strip()

        if choice == "1":
            tracker.add_expense()

        elif choice == "2":
            tracker.view_expenses()

        elif choice == "3":
            tracker.view_budget()

        elif choice == "4":
            print("Thank you for using the Expense Tracker!")
            break

        else:
            print("Invalid option. Please choose between 1 and 4.")


if __name__ == "__main__":
    main()
```
